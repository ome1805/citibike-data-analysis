```python
import duckdb
file_path = '202603-citibike-tripdata/202603-citibike-tripdata_1.csv'

query = f"""
SELECT 
    CASE 
        WHEN HOUR(started_at) BETWEEN 6 AND 11 THEN 'Morning (6am-11am)'
        WHEN HOUR(started_at) BETWEEN 12 AND 17 THEN 'Afternoon/Evening (12pm-5pm)'
        ELSE 'Night (6pm-5am)'
    END AS time_window,
    HOUR(started_at) AS hour_of_day,
    COUNT(*) AS total_trips
FROM read_csv_auto('{file_path}')
GROUP BY time_window, hour_of_day
ORDER BY time_window, total_trips DESC
"""

duckdb.sql(query).df()
```

```python
import duckdb
file_path = '202603-citibike-tripdata/202603-citibike-tripdata_1.csv'

query = f"""
WITH hourly_counts AS (
    SELECT 
        CASE 
            WHEN HOUR(started_at) BETWEEN 6 AND 11 THEN 'Morning (6am-11am)'
            WHEN HOUR(started_at) BETWEEN 12 AND 17 THEN 'Afternoon/Evening (12pm-5pm)'
            ELSE 'Night (6pm-5am)'
        END AS time_window,
        HOUR(started_at) AS hour_of_day,
        COUNT(*) AS total_trips
    FROM read_csv_auto('{file_path}')
    GROUP BY time_window, hour_of_day
)
SELECT 
    time_window,
    MAX_BY(hour_of_day, total_trips) AS peak_hour,
    MAX(total_trips) AS peak_trips
FROM hourly_counts
GROUP BY time_window
ORDER BY time_window
"""

duckdb.sql(query).df()
```
